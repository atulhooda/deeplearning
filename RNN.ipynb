Example: Weather Data

Suppose you want to predict tomorrow's weather. Every day you record 2 features:
Temperature (°C)
Humidity (%)
And you collect data for 3 days.

City 1
| Day   | Temperature | Humidity |
| ----- | ----------- | -------- |
| Day 1 | 30          | 70       |
| Day 2 | 31          | 68       |
| Day 3 | 29          | 75       |

City 2
| Day   | Temperature | Humidity |
| ----- | ----------- | -------- |
| Day 1 | 25          | 80       |
| Day 2 | 26          | 78       |
| Day 3 | 27          | 76       |


```python
weather_data = tf.constant([[[30,70],[31,68],[29,75]],[[25,80],[26,78],[27,76]]])
print(weather_data.shape)
```

2 → Cities (Batch Size)
3 → Days (Time Steps)
2 → Features (Temperature, Humidity)
