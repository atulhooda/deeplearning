Example: Weather Data

Suppose you want to predict tomorrow's weather. Every day you record 2 features:
Temperature (°C)
Humidity (%)
And you collect data for 3 days.

City 1
| Day   | Temperature | Humidity |
| ----- | ----------- | -------- |
| Day 1 | 30          | 70       |
| Day 2 | 31          | 68       |
| Day 3 | 29          | 75       |

City 2
| Day   | Temperature | Humidity |
| ----- | ----------- | -------- |
| Day 1 | 25          | 80       |
| Day 2 | 26          | 78       |
| Day 3 | 27          | 76       |


```python
weather_data = tf.constant([[[30,70],[31,68],[29,75]],[[25,80],[26,78],[27,76]]])
print(weather_data.shape)
```

2 → Cities (Batch Size)
3 → Days (Time Steps)
2 → Features (Temperature, Humidity)

```python
model = tf.keras.Sequential([
    tf.keras.layers.SimpleRNN(16, input_shape=(5, 2))
])

model.summary()
```

```python
#many to many RNN
import numpy as np

# 100 samples
X_train = np.random.rand(100, 5, 2)

# 100 target values
y_train = np.random.rand(100, 1)

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(5,2)),
    tf.keras.layers.SimpleRNN(16),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.fit(X_train, y_train, epochs=10)
```

```python
#gradient scaling
optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001,
    clipnorm=1.0
)

model.compile(
    optimizer=optimizer,
    loss="mse"
)
```

```python
optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001,
    clipvalue=0.5
)
```

```python
#bidirectional RNN
model = tf.keras.Sequential([
    tf.keras.layers.Bidirectional(
        tf.keras.layers.SimpleRNN(32),
        input_shape=(5,2)
    )
])

model.summary()
```

```python
#deep stacked RNN
model = tf.keras.Sequential([
    tf.keras.layers.SimpleRNN(
        32,
        return_sequences=True,
        input_shape=(5,2)
    ),
    tf.keras.layers.SimpleRNN(16),
    tf.keras.layers.Dense(1)
])

model.summary()
```

```python
#creating a LSTM
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(32, input_shape=(5,2)),
    tf.keras.layers.Dense(1)
])

model.summary()
```

```python
#return_sequences=True
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(
        32,
        return_sequences=True,
        input_shape=(5,2)
    )
])
```

```python
#Many-to-One LSTM
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(32, input_shape=(5,2)),
    tf.keras.layers.Dense(1)
])
```

```python
#Many-to-Many LSTM
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(
        32,
        return_sequences=True,
        input_shape=(5,2)
    ),
    tf.keras.layers.Dense(1)
])
```

```python
#stacked LSTM
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(
        64,
        return_sequences=True,
        input_shape=(5,2)
    ),
    tf.keras.layers.LSTM(32),
    tf.keras.layers.Dense(1)
])
```

```python
#Bidirectional LSTM
model = tf.keras.Sequential([
    tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(32),
        input_shape=(5,2)
    ),
    tf.keras.layers.Dense(1)
])
```

```python
#Complete Training Example
# Dummy data
X_train = np.random.rand(100, 5, 2)
y_train = np.random.rand(100, 1)

model = tf.keras.Sequential([
    tf.keras.layers.LSTM(32, input_shape=(5,2)),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=16
)
```
